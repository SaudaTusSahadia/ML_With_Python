file write

```python
# creating a new file and writing to it
with open("test.txt","w") as file:
    file.write("This is a sample text file.\n")
    file.write("It contains multiple lines of text.\n")
    file.write("This is the third line.\n")
```

```python
# writing on an existing file will overwrite the existing content of the file.
with open("test.txt","w") as file:
    file.write("Hello, World!\n")
    file.write("It contains multiple lines of text.\n")
    file.write("This is the third line.\n")
```

```python
# i dont want to overwrite the existing content of the file, so I will use the append mode.
with open("test.txt","a") as file:
    file.write("This is an additional line.\n")
    file.write("This line will be added to the end of the file.\n")
    file.write("Appending new content to the existing file.\n")
```

```python
strings = ['Hello, World!\n', 'This is a sample text file.\n', 'It contains multiple lines of text.\n', 'This is the third line.\n', 'This is an additional line.\n', 'This line will be added to the end of the file.\n', 'Appending new content to the existing file.\n']

with open("test2.txt","w") as file:
    file.writelines(strings)
```

practice 1:

get the text from sample.txt and count the following

number of lines
number of words
number of charatcers
and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("sample.txt","r") as file:
    strings = file.readlines()

    total_lines = len(strings)
    print("Total number of lines:", total_lines)

    number_of_words = sum(len(line.split()) for line in strings)
    print("Total number of words:", number_of_words)

    total_characters = reduce(lambda acc, line: acc + len(line), strings, 0)
    print("Total number of characters:", total_characters)

with open("counter_of_string.txt","w") as file:
    file.write(f"\nTotal number of lines: {total_lines}\n")
    file.write(f"Total number of words: {number_of_words}\n")
    file.write(f"Total number of characters: {total_characters}\n")

    print(strings)
```

write then read with one open

```python
with open("write_read.txt","w+") as file:
    file.write("hello world\n")

    print(file.tell())
    file.seek(0)
    print(file.tell())
    print(file.read(5))

    file.truncate(5)
    file.seek(0)
    print(file.read())
```
