```python
raining = True
Muddy = False

if raining==True:
    print("It's raining, don't forget your umbrella!")
    print("Stay dry and safe!")
    if Muddy==True:
        print("It's muddy outside, wear your boots!")
    else:
        print("It's not muddy, you can wear your regular shoes.")

else:
    print("It's not raining, enjoy the sunshine!")
    print("Have a great day!")
```

```python
taka = 100

if taka > 500:
    print("You have enough money to buy a new phone!")

elif taka > 400:
    print("You have enough money to buy a new pair of shoes!")

elif taka > 300:
    print("You have enough money to buy a new book!")

else:
    print("You might want to save up a bit more before making a purchase.")
```
