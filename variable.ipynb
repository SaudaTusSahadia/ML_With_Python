Codeforces Solutions

```python
inp = input()

# print(inp, type(inp))

numbers = inp.split()

# print(numbers, type(numbers))

x = int(numbers[0])
y = int(numbers[1])

# print(x, type(x))
# print(y, type(y))

last_digit_of_x = x % 10
last_digit_of_y = y % 10

sum = last_digit_of_x + last_digit_of_y

print("Sum of last digits:", sum)
```
