Max & Min Problem

```python
inp = input()
numbers = inp.split()
x = int(numbers[0])
y = int(numbers[1])
z = int(numbers[2])

min = x
max = x

#minimum
if y < min:
    min = y
if z < min:
    min = z


#maximum
if y > max:
    max = y
if z > max:
    max = z

print(min, max)
```

C. Even, Odd, Positive and Negative

```python
n = int(input())

inp = input()
numbers = inp.split()

positive_count = 0; negative_count = 0; even_count = 0; odd_count = 0

for i in range(n):

    x = int(numbers[i])
    # positive, negative, even, odd
    if x > 0:
        positive_count += 1
    elif x < 0:
        negative_count += 1
    if x % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even:", even_count)
print("Odd:", odd_count)
print("Positive:", positive_count)
print("Negative:", negative_count)
```

Q. Digits

```python
n = int(input())

for i in range(n):
    number = int(input())
    if number == 0:
        print(0)
        continue
    while number > 0:
        print(number%10, end=" ")
        number //= 10
    print()
```
